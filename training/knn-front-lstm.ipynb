```python
class CustomDataset(Dataset):
  def __init__(self, features, labels):
    self.features = features
    self.labels = labels

  def __len__(self):
    return len(self.features)

  def __getitem__(self, idx):
    return self.features[idx], self.labels[idx]
  
class FrontLSTMDataset(Dataset):
    def __init__(self, static_features, dynamic_features, labels, sequence_length=20):
        """
        static_features: shape (N, num_static_features) - features that don't change over time
        dynamic_features: shape (N, num_dynamic_features) - time-varying features
        labels: shape (N,) or (N, 1)
        sequence_length: number of time steps in each input sequence
        """
        self.sequence_length = sequence_length

        self.static_features = torch.tensor(static_features, dtype=torch.float32)
        self.dynamic_features = torch.tensor(dynamic_features, dtype=torch.float32)
        self.labels = torch.tensor(labels.to_numpy(), dtype=torch.float32)

        # Ensure labels have shape (N, 1)
        if self.labels.ndim == 1:
            self.labels = self.labels.unsqueeze(1)

    def __len__(self):
        return len(self.dynamic_features) - self.sequence_length

    def __getitem__(self, idx):
        # Static features for the target time step (don't change over sequence)
        x_static = self.static_features[idx + self.sequence_length]  # shape: (num_static_features,)
        
        # Dynamic features sequence
        x_dynamic = self.dynamic_features[idx : idx + self.sequence_length]  # shape: (seq_len, num_dynamic_features)
        
        # Target: next step
        y = self.labels[idx + self.sequence_length]
        
        return x_static, x_dynamic, y

class SequenceDataset(Dataset):
    def __init__(self, features, labels, sequence_length = 20):
        """
        features: shape (N, num_features)
        labels: shape (N,) or (N, 1)
        sequence_length: number of time steps in each input sequence
        """
        self.sequence_length = sequence_length

        self.features = torch.tensor(features, dtype=torch.float32)
        self.labels = torch.tensor(labels.to_numpy(), dtype=torch.float32)

        # Ensure labels have shape (N, 1)
        if self.labels.ndim == 1:
            self.labels = self.labels.unsqueeze(1)

    def __len__(self):
        return len(self.features) - self.sequence_length

    def __getitem__(self, idx):
        x = self.features[idx : idx + self.sequence_length]  # shape: (seq_len, num_features)
        y = self.labels[idx + self.sequence_length]          # target: next step
        return x, y
```

```python
import torch.nn as nn

class FrontLSTM(nn.Module):
    def __init__(self, input_size_sta, input_size_dyn, hidden_size, output_size, num_layers, drop_prob=0.5):
        super().__init__()

        self.hidden_size = hidden_size
        self.num_layers = num_layers

        self.fc_h = nn.Linear(input_size_sta, num_layers * hidden_size)
        self.fc_c = nn.Linear(input_size_sta, num_layers * hidden_size)
        if num_layers == 1:
            self.lstm = nn.LSTM(input_size_dyn, hidden_size, num_layers, batch_first=True)
        else:
            self.lstm = nn.LSTM(input_size_dyn, hidden_size, num_layers,
                                dropout=drop_prob, batch_first=True)
        self.dropout = nn.Dropout(drop_prob)
        self.fc = nn.Linear(hidden_size, output_size)
        self.act = nn.ReLU()

    def forward(self, x_s, x_d):
        # Set initial hidden and cell states
        h0 = self.act(self.fc_h(x_s))
        h0 = h0.reshape(h0.shape[0], self.num_layers, -1).transpose(0, 1).contiguous()

        c0 = self.act(self.fc_c(x_s))
        c0 = c0.reshape(c0.shape[0], self.num_layers, -1).transpose(0, 1).contiguous()

        # Forward propagate LSTM
        # out: tensor of shape (batch_size, seq_length, hidden_size)
        out, _ = self.lstm(x_d, (h0, c0))

        # Decode the hidden state of the last time step
        out = self.dropout(out[:, -1, :])
        out = self.fc(out)

        # use specific activation
        out = self.act(out)
        return out

class LSTMModel(nn.Module):
    def __init__(self, input_size, num_layers):
        super().__init__()

        self.lstm1 = nn.LSTM(input_size, 128, num_layers, batch_first=True, bidirectional=True, dropout=0.2 if num_layers > 1 else 0)
        self.dropout1 = nn.Dropout(0.5)
        self.relu1 = nn.ReLU()

        self.lstm2 = nn.LSTM(128*2, 64, num_layers, batch_first=True, bidirectional=False, dropout=0.2 if num_layers > 1 else 0)
        self.dropout2 = nn.Dropout(0.5)
        self.relu2 = nn.ReLU()

        # self.lstm3 = nn.LSTM(64, 32, num_layers, batch_first=True, dropout=0.2 if num_layers > 1 else 0)
        # self.dropout3 = nn.Dropout(0.5)
        # self.relu3 = nn.ReLU()

        self.fc = nn.Linear(64, 1)
        
    #     # Initialize weights properly
    #     self._init_weights()

    # def _init_weights(self):
    #     for name, param in self.named_parameters():       
    #         if 'weight_ih' in name:
    #             nn.init.xavier_uniform_(param.data)
    #         elif 'weight_hh' in name:
    #             nn.init.orthogonal_(param.data)
    #         elif 'bias' in name:
    #             param.data.fill_(0)
    #         elif 'fc.weight' in name:
    #             nn.init.xavier_uniform_(param.data)
    #         elif 'fc.bias' in name:
    #             param.data.fill_(0)

    def forward(self, x):
        out, _ = self.lstm1(x)
        out = self.dropout1(out)
        out = self.relu1(out)

        out, _ = self.lstm2(out)
        out = self.dropout2(out)
        out = self.relu2(out)

        # out, _ = self.lstm3(out)
        # out = self.dropout3(out)
        # out = self.relu3(out)

        out = out[:, -1, :]  # Last time step
        out = self.fc(out)
        return out
```

```python
class LogRMSELoss(nn.Module):
    def __init__(self, eps=1e-8):
        super().__init__()
        self.eps = eps  # Increased epsilon for better stability

    def forward(self, pred, target):
        # Ensure positive values and handle edge cases
        pred = torch.clamp(pred, min=self.eps)
        target = torch.clamp(target, min=self.eps)
        
        # Check for NaN or inf values
        if torch.isnan(pred).any() or torch.isnan(target).any():
            return torch.tensor(float('inf'), device=pred.device)

        # Compute log RMSE with additional stability checks
        try:
            log_pred = torch.log(pred + self.eps)
            log_target = torch.log(target + self.eps)
            loss = torch.sqrt(torch.mean((log_pred - log_target) ** 2) + self.eps)
            return loss
        except:
            return torch.tensor(float('inf'), device=pred.device)
        
class RMSELoss(torch.nn.Module):
    def __init__(self):
        super(RMSELoss,self).__init__()

    def forward(self,x,y):
        criterion = nn.MSELoss()
        loss = torch.sqrt(criterion(x, y))
        return loss
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.amp import autocast, GradScaler
from sklearn.metrics import r2_score, mean_absolute_error
import wandb
import os
import numpy as np

criterion = RMSELoss()
optimizer = optim.Adam(model.parameters(), lr=0.003, weight_decay=1e-4)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Mixed precision training setup
scaler = GradScaler()

# Learning rate scheduler for faster convergence
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5)

# Set up model saving directory
os.makedirs("checkpoints", exist_ok=True)

# --- Early Stopping Setup ---
best_val_loss = float('inf')
patience = 75
patience_counter = 0
val_freq = 5  # Validate every 5 epochs for speed

# Training Loop
for epoch in range(500):
    model.train()
    train_loss = 0.0
    train_predictions = []
    train_targets = []
    
    for batch_idx, (x_static, x_dynamic, yb) in enumerate(train_loader):
        x_static, x_dynamic, yb = x_static.to(device, non_blocking=True), x_dynamic.to(device, non_blocking=True), yb.to(device, non_blocking=True)

        optimizer.zero_grad()
        
        # Mixed precision forward pass - specify device_type
        with autocast(device_type='cuda'):
            pred = model(x_static, x_dynamic)
            loss = criterion(pred, yb)
        
        # Mixed precision backward pass
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        
        # Gradient clipping
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        
        scaler.step(optimizer)
        scaler.update()
        
        train_loss += loss.item()
        
        # Collect training predictions and targets for metrics
        train_predictions.append(pred.detach().cpu().numpy())
        train_targets.append(yb.cpu().numpy())

    train_loss /= len(train_loader)
    
    # Calculate training MAE and R2
    train_predictions = np.concatenate(train_predictions)
    train_targets = np.concatenate(train_targets)
    train_mae = mean_absolute_error(train_targets, train_predictions)
    train_r2 = r2_score(train_targets, train_predictions)
    
    # Validation every val_freq epochs
    if epoch % val_freq == 0 or epoch == 199:
        model.eval()
        val_loss = 0.0
        val_predictions = []
        val_targets = []
        
        with torch.no_grad():
            for x_static, x_dynamic, yb in val_loader:
                x_static, x_dynamic, yb = x_static.to(device, non_blocking=True), x_dynamic.to(device, non_blocking=True), yb.to(device, non_blocking=True)
                with autocast(device_type='cuda'):
                    pred = model(x_static, x_dynamic)
                    loss = criterion(pred, yb)
                if not torch.isnan(loss):
                    val_loss += loss.item()
                
                # Collect predictions and targets for metrics
                val_predictions.append(pred.cpu().numpy())
                val_targets.append(yb.cpu().numpy())

        val_loss /= len(val_loader)
        scheduler.step(val_loss)  # Update learning rate
        
        # Calculate MAE and R2 scores
        val_predictions = np.concatenate(val_predictions)
        val_targets = np.concatenate(val_targets)
        val_mae = mean_absolute_error(val_targets, val_predictions)
        val_r2 = r2_score(val_targets, val_predictions)
        
        print(f"Epoch {epoch+1}, Train Loss: {train_loss:.4f}, Train MAE: {train_mae:.4f}, Train R2: {train_r2:.4f}, Val Loss: {val_loss:.4f}, Val MAE: {val_mae:.4f}, Val R2: {val_r2:.4f}, LR: {optimizer.param_groups[0]['lr']:.6f}")
        wandb.log({
            "epoch": epoch+1, 
            "train_loss": train_loss, 
            "train_mae": train_mae,
            "train_r2": train_r2,
            "val_loss": val_loss, 
            "val_mae": val_mae,
            "val_r2": val_r2,
            "learning_rate": optimizer.param_groups[0]['lr']
        })

        # --- Checkpoint saving if improved ---
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            patience_counter = 0
            torch.save(model.state_dict(), f"checkpoints/best_frontlstm_fast.pt")
            print(f"✅ Saved checkpoint at epoch {epoch+1} (val_loss={val_loss:.4f})")
        else:
            patience_counter += 1
            print(f"⚠️  No improvement. Patience: {patience_counter}/{patience}")
            if patience_counter >= patience:
                print("⛔ Early stopping triggered")
                break
    else:
        print(f"Epoch {epoch+1}, Train Loss: {train_loss:.4f}, Train MAE: {train_mae:.4f}, Train R2: {train_r2:.4f}")
        wandb.log({"epoch": epoch+1, "train_loss": train_loss, "train_mae": train_mae, "train_r2": train_r2})
```
